# Runtime latency-budget control (PI)

A PI controller adjusts the routing threshold τ frame-by-frame to track a
time-varying latency budget L*(t), as the scene condition drifts.

- **Live on-device demo (Fig. 8)** — closed loop on the RTX 3090, measured per frame.
- **Scenario figures** — step / sawtooth budgets across condition-drift sequences.
- **Gain sensitivity** — robustness of tracking to the PI gains.
